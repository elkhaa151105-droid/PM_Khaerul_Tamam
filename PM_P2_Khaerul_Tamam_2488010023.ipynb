```python
import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]}  # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']]   # fitur
y = df['harga']                     # label
```

```python
# Label numerik kontinu -> REGRESI
print(df['harga'].tolist())
# Label kategori -> KLASIFIKASI
df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']
```

```python

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), ' Uji:', len(X_test))
```

3.  Untuk kasus 'memprediksi jumlah pengunjung toko besok', tentukan jenis pembelajaran, subkategori, fitur, dan label.

Kasus “memprediksi jumlah pengunjung toko besok” termasuk Supervised Learning dengan subkategori regresi. Fitur yang dapat digunakan berupa jumlah pengunjung pada hari-hari sebelumnya, hari dalam minggu, adanya promosi, kondisi cuaca, dan informasi lain yang berhubungan dengan jumlah pengunjung. Label yang diprediksi adalah jumlah pengunjung toko besok. Kasus ini termasuk regresi karena hasil yang ingin diprediksi berupa nilai numerik, yaitu jumlah orang yang datang ke toko.


E. Refleksi
1. Supervised learning adalah pembelajaran mesin yang menggunakan data yang sudah memiliki label atau jawaban sebagai bahan untuk melatih model. Contohnya adalah memprediksi apakah siswa lulus atau tidak berdasarkan jam belajar, karena data sebelumnya sudah memiliki hasil lulus atau tidak lulus. Unsupervised learning adalah pembelajaran mesin yang menggunakan data tanpa label atau jawaban yang sudah ditentukan. Model akan mencari pola atau kelompok tertentu dari data tersebut. Contohnya adalah mengelompokkan pelanggan berdasarkan kebiasaan belanja tanpa menentukan terlebih dahulu kelompok pelanggannya. Reinforcement learning adalah pembelajaran mesin dengan cara memberikan penghargaan (reward) atau hukuman (penalty) berdasarkan tindakan yang dilakukan oleh model. Model akan belajar memilih tindakan yang menghasilkan hasil terbaik melalui proses percobaan dan kesalahan. Contohnya adalah komputer yang belajar bermain game dengan mendapatkan poin ketika melakukan tindakan yang benar.

Pembagian data latih dan data uji penting karena data latih digunakan untuk melatih model, sedangkan data uji digunakan untuk mengetahui kemampuan model dalam memprediksi data yang belum pernah dilihat sebelumnya. Dengan adanya pembagian ini, kita dapat menilai apakah model benar-benar mampu mengenali pola atau hanya menghafal data yang digunakan saat latihan. 3. Alur kerja Machine Learning (ML) disebut iteratif karena proses pembuatan model tidak hanya dilakukan satu kali, tetapi dilakukan secara berulang-ulang untuk mendapatkan hasil yang lebih baik. Setelah model dilatih dan diuji, hasilnya dievaluasi untuk mengetahui kekurangan atau kesalahannya. Jika hasilnya belum sesuai, data, fitur, algoritma, atau parameter model dapat diperbaiki dan proses pelatihan dilakukan kembali. Proses tersebut terus diulang sampai diperoleh model dengan hasil yang sesuai dan memiliki kinerja yang baik.
